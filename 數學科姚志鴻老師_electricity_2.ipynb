###<font color = '#22a6b3'>8. 處理缺漏的資料</font>

折線圖的整體趨勢：

臺灣的供電需求量和總發電量整體呈現上升趨勢，表示隨著時間推移，臺灣的用電需求不斷增加，而發電量也隨之提高以滿足需求。
供電需求量和總發電量的變化趨勢非常接近，幾乎重疊在一起，這表示臺灣的電力供應與需求大致平衡。
「呈現直角」的圖形特徵，代表...

資料在短時間內發生劇烈變化。
在特定年份，供電需求量或總發電量突然大幅增加或減少。
「呈現直角」的圖形特徵，有可能跟資料的起始記錄時間有關。

如果在 2000 年以前，臺灣並沒有系統性地記錄電力供需資料，那麼在 2000 年開始記錄時，就會出現一個突然的數據變化，在圖表上就會呈現出「直角」的形狀。

舉例來說：

假設在 2000 年以前，臺灣的電力供需資料並未完整記錄或公開。到了 2000 年，政府開始建立電力資料庫，並開始記錄每年的供電需求量和總發電量。那麼，在繪製折線圖時，就會出現以下情況：

1999 年以前的資料缺失或不完整，導致折線圖無法顯示該時期的數據。
2000 年開始，出現完整的電力供需資料，因此折線圖會從 2000 年開始顯示數據。
由於 1999 年和 2000 年之間的資料存在巨大差異，因此折線圖會在這兩個年份之間出現一個急劇的轉折，形成「直角」。

###<font color = '#22a6b3'>6. 比較其他國家的供電與發電狀況</font>

**除了臺灣，猜猜看其他國家的發電量與供電需求量是否也會「幾乎完全一致」？**

**請選擇以下五個國家：日本（JPN）、美國（USA）、加拿大（CAN）、德國（DEU）、法國（FRA），並針對 2000 年以後的資料，繪製五張對應的折線圖（每國一張）**

<font color = '#1e3799'>[提示]</font> 你可以試著用迴圈搭配列表，讓程式幫你一口氣畫完五張圖！

```python
countries = ['JPN', 'USA', 'CAN', 'DEU', 'FRA']
import plotly.express as px
for country in countries:
    country_data = df[(df['iso_code'] == country) & (df['year'] >= 2000)]
    fig = px.line(country_data,
                  x='year',
                  y=['electricity_demand', 'electricity_generation'],
                  title=f'{country} 逐年供電需求量與總發電量變化圖')
    fig.show()
```
